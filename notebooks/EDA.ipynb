```python
import pandas as pd
import plotly.express as px
```

```python
# Lê os dados uma única vez
car_data = pd.read_csv('../vehicles.csv')
car_data.head()
```

```python
# Visão geral: tipos, valores ausentes e estatísticas
car_data.info()
print('\nValores ausentes por coluna:')
print(car_data.isna().sum())
car_data.describe()
```

```python
fig = px.histogram(car_data, x='odometer', labels={'odometer': 'Odômetro'})
fig.update_layout(yaxis_title='Quantidade de anúncios')
fig.show()
```

```python
fig = px.scatter(car_data, x='odometer', y='price',
                 labels={'odometer': 'Odômetro', 'price': 'Preço (US$)'})
fig.show()
```
